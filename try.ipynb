```python
import numpy as np
def f(x):
    return x**2
def gradient(x):
    return 2*x
x = 10.0
learning_rate = 0.1
for i in range(100):
    grad = gradient(x)
    x = x - learning_rate * grad
    print(f"Iteration {i+1}: x = {x}, f(x) = {f(x)}")
```

```python
x = tf.Variable(10.0)
with tf.GradientTape() as tape:
    y = x**2
grad = tape.gradient(y, x)
learning_rate = 0.1
print(f"Gradient: {grad.numpy()}")
```
